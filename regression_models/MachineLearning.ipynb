```python
target_column = "Y house price of unit area"
unwanted_columns = ["No","X1 transaction date"]

data_clean=data.drop(columns=unwanted_columns)
```

```python
data_clean
```

```python
features = data_clean.drop(columns=target_column)
target = data_clean[target_column]
```

```python
features
```

```python
target
```

## Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=30)
```

```python
len(X_train) == len(y_train)  # Check if X_train and y_train have the same number of rows
```

```python
len(X_test) == len(y_test)  # Check if X_test and y_test have the same number of rows
```

## Pre-Processing

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler
```

```python
scaled_data = scaler.fit_transform(features)  #Scales all the features
```

# Model Building

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_regression = LinearRegression()
```

```python
linear_regression.fit(X_train, y_train)
```

```python
linear_regression.coef_  # Weight for each feature
```

```python
linear_regression.intercept_  # Intercept Value
```

# Train Performance

```python
y_pred_train = linear_regression.predict(X_train)  # Predict the prices using train data
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mse = mean_squared_error(y_train, y_pred_train)
mse
```

```python
import numpy as np
```

```python
np.sqrt(mse)
```

# Test Performance

```python
y_pred_test = linear_regression.predict(X_test)  # Predict the prices using test data
```

```python
mse_test = mean_squared_error(y_test, y_pred_test)
mse_test
```

```python
np.sqrt(mse_test)
```
