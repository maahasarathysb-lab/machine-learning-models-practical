# slicing 1d Arrays

```python
# find the top 5 performing agents or names by sales
# sort the sales in desc
#slice the top
#index it against the agents or name 
```

```python
import numpy as np

agents = np.array([
    "Arun", "Bala", "Charan", "Deepak",
    "Esha", "Farhan", "Gokul", "Hari"
])

sales = np.array([450, 800, 600, 1200, 700, 950, 500, 1100])
```

```python
sorted_index = np.argsort(sales)[::-1]
sorted_index
```

```python
top5_index = sorted_index[:5]
top5_index
```

```python
top5_agents = agents[top5_index]
print(top5_agents)
```

```python
top5_sales = sales[top5_index]
print(top5_sales)
```

```python
#shotest and fasted approach
top5_index = np.argsort(sales)[::-1][:5]

print(agents[top5_index])
print(sales[top5_index])
```

```python
last5_index = np.argsort(sales)[:5]

print(agents[last5_index])
print(sales[last5_index])
```
