### 6.1 — BYJU'S Student Score Analysis

```python
test_scores = np.array([72, 85, 60, 91, 78, 65, 88, 54, 95, 70])

print("Student Scores  :", test_scores)
print("Mean Score      :", np.mean(test_scores))
print("Median Score    :", np.median(test_scores))
print("Std Deviation   :", np.round(np.std(test_scores), 2))
print("Min Score       :", np.min(test_scores))
print("Max Score       :", np.max(test_scores))
print("Total Sum       :", np.sum(test_scores))
print("25th Percentile :", np.percentile(test_scores, 25))
print("75th Percentile :", np.percentile(test_scores, 75))
```

## vstack / hstack — Combine Zone-wise Sales

```python
north_sales = np.array([100, 200, 300])
south_sales = np.array([150, 250, 350])

combined = np.vstack([north_sales, south_sales])   # rows stacked
side_by  = np.hstack([north_sales, south_sales])   # columns joined

print("North Sales:", north_sales)
print("South Sales:", south_sales)
print("\nvstack (2 rows):\n", combined)
print("hstack (1 row) :", side_by)
```
