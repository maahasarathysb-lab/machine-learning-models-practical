```python
import numpy as np

print("="*70)
print("NUMPY BROADCASTING - COMPLETE CHEAT SHEET")
print("="*70)

# ----------------------------------------------------
# Rule 1: Same Shape
# ----------------------------------------------------
print("\nRule 1: Same Shape")
a = np.array([1,2,3])
b = np.array([10,20,30])

print("Shape:", a.shape, b.shape)
print(a+b)

# ----------------------------------------------------
# Rule 2: Scalar Broadcasting
# ----------------------------------------------------
print("\nRule 2: Scalar Broadcasting")
a = np.array([[1,2],[3,4]])

print(a)
print(a+10)

# ----------------------------------------------------
# Rule 3: Row Broadcasting
# ----------------------------------------------------
print("\nRule 3: Row Broadcasting")

a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])

print("Shapes:", a.shape, b.shape)
print(a+b)

# ----------------------------------------------------
# Rule 4: Column Broadcasting
# ----------------------------------------------------
print("\nRule 4: Column Broadcasting")

a = np.array([[1],
              [2],
              [3]])

b = np.array([10,20,30])

print("Shapes:", a.shape, b.shape)
print(a+b)

# ----------------------------------------------------
# Rule 5: 2D + 2D
# ----------------------------------------------------
print("\nRule 5: Same 2D Shape")

a = np.array([[1,2],
              [3,4]])

b = np.array([[10,20],
              [30,40]])

print(a+b)

# ----------------------------------------------------
# Rule 6: Broadcasting Error
# ----------------------------------------------------
print("\nRule 6: Error Example")

a = np.array([1,2,3])
b = np.array([10,20])

print("Shapes:", a.shape, b.shape)

try:
    print(a+b)
except Exception as e:
    print("Error:", e)

# ----------------------------------------------------
# Rule 7: Reshape to Fix Broadcasting
# ----------------------------------------------------
print("\nRule 7: Fix Using reshape")

a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20]).reshape(2,1)

print("Shapes:", a.shape, b.shape)
print(a+b)

# ----------------------------------------------------
# Rule 8: Real World Example (Monthly Bonus)
# ----------------------------------------------------
print("\nRule 8: Employee Monthly Bonus")

sales = np.array([
    [100,200,300],
    [150,250,350],
    [180,280,380]
])

bonus = np.array([10,20,30])

print("Original Sales")
print(sales)

print("\nBonus")
print(bonus)

print("\nAfter Bonus")
print(sales+bonus)

# ----------------------------------------------------
# Rule 9: Salary Increment
# ----------------------------------------------------
print("\nRule 9: Salary Increment")

salary = np.array([30000,40000,50000])

print(salary+5000)

# ----------------------------------------------------
# Rule 10: Broadcasting Conditions
# ----------------------------------------------------
print("\nBroadcasting Rules")
print("1. Compare shapes from RIGHT to LEFT")
print("2. Dimensions must be Equal OR one of them must be 1")
print("3. Missing dimension is treated as 1")
print("4. Otherwise -> ValueError")

print("\nExamples")
print("(3,)  + (3,)   -> Works")
print("(3,3) + (3,)   -> Works")
print("(3,1) + (3,)   -> Works")
print("(2,3) + (2,1)  -> Works")
print("(3,)  + (2,)   -> Error")
print("(2,3) + (2,)   -> Error")
```
