```python
import re

text = """
my aadhaar card number is 9495 8589 4589
and credit card no 7894-7894-1896-5000
9894679880
"""

# Aadhaar: show only center 4 digits
text = re.sub(
    r"(\d{4})\s(\d{4})\s(\d{4})",
    r"XXXX \2 XXXX",
    text
)

# Credit card: hide first 12 digits
text = re.sub(
    r"(\d{4})-\d{4}-\d{4}-\d{4}",
    r"\1-XXXX-XXXX-XXXX",
    text
)

# Mobile: hide center 4 digits
text = re.sub(
    r"(\d{3})\d{4}(\d{3})",
    r"\1XXXX\2",
    text
)

print(text)
```
