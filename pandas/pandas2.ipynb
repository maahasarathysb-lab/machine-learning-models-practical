```python
import pandas as pd
import numpy as np

amount = np.array([500, 700, 900, 1000, 7000, 5000, 2000])
weekdays = np.array(['monday','tuesday','wednesday','thursday','friday','saturday','sunday'])

sales = pd.Series(
    data=amount,
    index=weekdays,
    name="sales",
    dtype=np.int16
)

print(sales)
```

```python
dir(sales)  # attributes and methods of series object
```

```python
sales.values #as a numpy array
```

```python
sales.index # extract index of the series objects
```

```python
sales.isna() # to identify null values 
```

```python
sales.isna().any() # outputs ture if any of the aavlues is true 
```

```python
sales.isna().all() # outputs ture if all of the aavlues is true 
```
