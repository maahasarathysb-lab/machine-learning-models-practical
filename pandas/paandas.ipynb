```python
# find the averaage profit of state
sales_data.groupby("Region")["Profit"].mean()
```

```python
# find the averaage profit and sales  of state
sales_data.groupby("Region")[["Profit", "Sales","Discount"]].mean()
```

```python
#find the average product , total_sales and total orders by region
sales_agg = sales_data.groupby("Region").agg({
    "Profit": "mean",
    "Sales": "sum",
    "Order ID": "count"
})
sales_agg
```

```python
## rename column to match the values in the dataframe
sales_agg.rename?
```

```python
sales_agg.rename(mapper={"Profit":"avg profit",
                        "Sales":"total_sales",
                        "Order ID":"total_order"})
```

```python
# find the averafe shipping cost and total order priority
#rename the columns that align with the values in the result
sales_data.groupby(["Region", "Order Priority","Shipping Cost"]).size()
```

```python
sales_data = sales_data.drop(columns=("Row ID"))
```

```python
sales_data.info()
```

```python
sales_data['Ship Date'] - sales_data['Order Date']
```

```python
sales_data['Ship Time'] = sales_data['Ship Date'] - sales_data['Order Date']
```

```python
sales_data
```

```python
# create a average ship time by order priroity
average_shiptime = sales_data.groupby("Order Priority").agg({
    "Ship Time": "mean"
})

average_shiptime
```

```python
sales_data.info()
```

```python
sales_data['ship time (days)'] = sales_data['Ship Time'].map(lambda timedelta: timedelta.days)
```

```python
sales_data.head()
```

```python
# create a average ship time by order priroity
average_shiptime = sales_data.groupby("Order Priority").agg({
    "ship time (days)": "mean"
})

average_shiptime
```

```python
# the company has a policy that every order that has a ship time of more than 2 days to be 
# tagged as 'delay', 'no delay' otherwise'
# add a new coloumn 'delay status' that should hold aonly two values 'delay'/'no delay'
# based on the ship time (days) column
sales_data['delay status']=sales_data['ship time (days)'].map(lambda timedelta: 'delay' if timedelta > 2 else 'no delay')
```

```python
sales_data.tail()
```

```python
np.where(sales_data['ship time (days)']> 2,"delay", "no delay")
```

```python
# find out if there is any association between container and delay status
# it is probable that large container shipment causes delay in shipping 
pd.crosstab(sales_data['Product Container'],sales_data['delay status'])
```

```python
# Calculate delay percentage per container type and sort from highest to lowest
delay_rates = (
    pd.crosstab(
        sales_data['Product Container'],
        sales_data['delay status'],
        normalize='columns',
    )['delay']
    * 100
)

delay_rates.sort_values(ascending=False)
```
