```python
import pandas as pd
import numpy as np

amount = np.array([500, 700, 900, 1000, 7000, 5000, 2000])
weekdays = np.array(['monday','tuesday','wednesday','thursday','friday','saturday','sunday'])

sales = pd.Series(
    data=amount,
    index=weekdays,
    name="sales",
    dtype=np.int16
)

print(sales)
```

```python
def discount(sales):
    if sales >= 1000:
        discount_amount = sales * 0.10   # 10% discount
    elif sales >= 500:
        discount_amount = sales * 0.05   # 5% discount
    else:
        discount_amount = 0

    remaining_amount = sales - discount_amount
    return discount_amount, remaining_amount
```

```python
sales.map(discount)
```
