# supervised learning  

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("https://raw.githubusercontent.com/raveendran2862/dataset/refs/heads/main/Salary_Data_1.csv")
df
```

```python
df.shape
```

```python
df.dropna(axis = 0,inplace = True)
df
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace= True)
```

```python
df.shape
```

```python
df.head()
```

#### x = input (experiance) , y =  target salary 

```python
x = df.iloc[:,4].values.reshape(-1,1)
x
```

```python
y = df.iloc[:,-1]
y
```

```python
plt.scatter(x,y)
plt.xlabel("years of exoeriance ")
plt.ylabel("salary")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr
```

```python
lr.fit(x,y)
```

```python
x.max()
```

```python
lr.predict([[27]])
```

```python
y_pred = lr.predict(x)
y_pred
```

```python
from sklearn.metrics import r2_score
r2_score(y,y_pred)
```

```python
plt.scatter(x, y)
plt.plot(x, y_pred, color='red')
plt.xlabel("years of experience")
plt.ylabel("salary")
plt.show()
```

# excercise

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
fd = pd.read_csv("https://raw.githubusercontent.com/raveendran2862/dataset/refs/heads/main/housing_dataset.csv")
fd
```

```python
fd.isnull().sum()
```

```python
fd.duplicated().sum()
```

```python
# Features (SqFt, Bedrooms, Bathrooms)
x = fd.iloc[:, [2, 3, 4]].values
x

```

```python
# Target (Price)
y = fd.iloc[:, 1].values
y
```

```python
plt.scatter(fd.iloc[:,2], y)  # SqFt vs Price
plt.show()

plt.scatter(fd.iloc[:,3], y)  # Bedrooms vs Price
plt.show()

plt.scatter(fd.iloc[:,4], y)  # Bathrooms vs Price
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr
```

```python
lr.fit(x,y)
```

```python
x.max()
```

```python
lr.predict([[2780,5,5]])
```

```python
y_ppred = lr.predict(x)
y_ppred
```

```python
from sklearn.metrics import r2_score
r2_score(y,y_ppred)
```

```python
plt.scatter(fd.iloc[:,2], y)# SqFt vs Price
plt.plot(x, y_ppred, color='red')
plt.show()

plt.scatter(fd.iloc[:,3], y)  # Bedrooms vs Price
plt.plot(x, y_ppred, color='red')

plt.show()

plt.scatter(fd.iloc[:,4], y)  # Bathrooms vs Price
plt.plot(x, y_ppred, color='red')
plt.show()
```

# chatgpt approach

```python
import numpy as np

# -------- SqFt vs Price --------
x_sqft = fd.iloc[:, 2].values
sorted_idx = np.argsort(x_sqft)

plt.scatter(x_sqft, y)
plt.plot(x_sqft[sorted_idx], y_ppred[sorted_idx], color='red')
plt.xlabel("SqFt")
plt.ylabel("Price")
plt.show()


# -------- Bedrooms vs Price --------
x_bed = fd.iloc[:, 3].values
sorted_idx = np.argsort(x_bed)

plt.scatter(x_bed, y)
plt.plot(x_bed[sorted_idx], y_ppred[sorted_idx], color='red')
plt.xlabel("Bedrooms")
plt.ylabel("Price")
plt.show()


# -------- Bathrooms vs Price --------
x_bath = fd.iloc[:, 4].values
sorted_idx = np.argsort(x_bath)

plt.scatter(x_bath, y)
plt.plot(x_bath[sorted_idx], y_ppred[sorted_idx], color='red')
plt.xlabel("Bathrooms")
plt.ylabel("Price")
plt.show()
```
