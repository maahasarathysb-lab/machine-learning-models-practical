```python
import nltk
import re
import pandas as pd

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
cleaned_reviews = []
targets = []

with open("amazon_cells_labelled.txt", "r", encoding="utf-8") as file:
    for line in file:
        parts = line.strip().split("\t")
        
        if len(parts) != 2:
            continue  # skip bad lines
        
        review, target = parts
        cleaned_reviews.append(review)
        targets.append(target)

print("Loaded:", len(cleaned_reviews))
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')   # ← this is the new requirement
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()
    
    # remove punctuation & numbers
    text = re.sub(r'[^a-z\s]', '', text)
    
    # tokenize
    words = word_tokenize(text)
    
    # remove stopwords
    words = [w for w in words if w not in stop_words]
    
    # lemmatize
    words = [lemmatizer.lemmatize(w, pos='v') for w in words]
    
    return " ".join(words)   # return as string (IMPORTANT)
```

```python
processed_reviews = [preprocess(review) for review in cleaned_reviews]

print(processed_reviews[:5])  # check sample
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    processed_reviews,
    targets,
    test_size=0.2,
    stratify=targets,
    random_state=42
)
```

```python
vectorizer = TfidfVectorizer()

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
model = LogisticRegression()

model.fit(X_train_vec, y_train)
```

```python
y_pred = model.predict(X_test_vec)

print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
new_review = ["This movie was absolutely amazing"]

new_review_processed = [preprocess(new_review[0])]
new_review_vec = vectorizer.transform(new_review_processed)

prediction = model.predict(new_review_vec)

print("Prediction:", prediction[0])
```

```python
y_pred = model.predict(X_test_vec)

print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
for i in range(5):
    print("Review:", X_test[i])
    print("Actual:", y_test[i])
    print("Predicted:", y_pred[i])
    print("------")
```

```python
test_reviews = [
    "Amazing movie, I loved it",
    "Worst movie ever",
    "It was okay, not great",
    "Fantastic acting and story"
]

for review in test_reviews:
    processed = [preprocess(review)]
    vec = vectorizer.transform(processed)
    pred = model.predict(vec)
    
    print(review, "→", pred[0])
```
