## line plot

```python
dowjones = sns.load_dataset("dowjones")
dowjones
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create figure and axes with custom size
fig, ax = plt.subplots(figsize=(15, 8))

# Plot the lineplot on the axes
sns.lineplot(data=dowjones, x="Date", y="Price", ax=ax)

# Add labels and title
ax.set_xlabel("Date")
ax.set_ylabel("Price")
ax.set_title("Dow Jones Price Over Time")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Make sure 'Date' is datetime
dowjones['Date'] = pd.to_datetime(dowjones['Date'])

# Calculate price change for color coding
dowjones['Change'] = dowjones['Price'].diff()  # current price - previous price

# Create figure
fig, ax = plt.subplots(figsize=(14, 9))

# Plot each segment individually with color
for i in range(1, len(dowjones)):
    color = 'green' if dowjones['Change'].iloc[i] >= 0 else 'red'
    ax.plot(dowjones['Date'].iloc[i-1:i+1], dowjones['Price'].iloc[i-1:i+1], color=color)

# Labels & title
ax.set_xlabel("Date")
ax.set_ylabel("Price")
ax.set_title("Dow Jones Price Movement (Green = Up, Red = Down)")

# Optional: rotate x-axis labels for readability
plt.xticks(rotation=45)

plt.show()
```

## heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Load dataset
flights = sns.load_dataset('flights')

# Pivot the data to create a matrix
flights_pivot = flights.pivot(index='month', columns='year', values='passengers')

# Create heatmap
sns.heatmap(flights_pivot, annot=True, fmt="d", cmap="YlGnBu")

plt.title("Flights Heatmap (Passengers per Month and Year)")
plt.xlabel("Year")
plt.ylabel("Month")
plt.show()
```

```python
sns.heatmap(flights_pivot, cmap="magma", cbar=True, square=True)

plt.title("Flights Heatmap - Minimalist Style")
plt.show()
```

```python
import pandas as pd
import seaborn as sns

# Load dataset
flights = sns.load_dataset('flights')

# Pivot table with explicit observed parameter
flights_pivot = pd.pivot_table(
    data=flights,
    index='month',       # rows
    columns='year',      # columns
    values='passengers',
    observed=False       # prevents FutureWarning
)

flights_pivot
```

```python
sns.heatmap(data=flights_pivot)
           
```

```python
sns.heatmap(data=flights_pivot,annot=True,fmt=".0f")
```

```python

plt.subplots(figsize=(15.0, 6.0))
sns.heatmap(data=flights_pivot,annot=True,fmt=".0f",cmap="Blues",linewidths=2,linecolor="white")
```
