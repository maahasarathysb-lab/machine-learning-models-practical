````python
import matplotlib.pyplot as plt
import seaborn as sns
````

## Bar Plot

bar plot is tupically used to find the association between a categorical and a continous numerical variable

````python
import pandas as pd
sales = pd.read_excel('sales_data_processed.xlsx')
````

````python
sales.head()
````

````python
sns.barplot?
````

````python
#each bar represents the average  shipping cost.
#vertical(black) line on each  bar represents the standard deviation
plt.subplots(figsize=(10,5))

sns.barplot(data=sales, x="delay status",y="Shipping Cost", palette = ["red","green"])
plt.title("Average shipping cost by delivery status",
         fontweight='bold',
         fontsize=15,
         color='Red')
plt.xlabel("Delay status", fontweight='bold', fontsize=12, color = 'black')
plt.ylabel("Delay status", fontweight='bold',  fontsize=12, color = 'black')
````

## count plot
count plot is used count the occurabce in each category of the categorical cariable

````python
# Plot the count of delay and no delay orders
plt.subplots(figsize=(10, 5))
sns.countplot(data=sales, x="delay status")
plt.title(
    "Count of Delay and No Delay Orders",
    fontsize=14,
    fontweight="bold",
    color="green",
)
````

````python
# Plot the count of delay and no delay orders
plt.subplots(figsize=(10, 5))
sns.countplot(data=sales, x="delay status",
             hue="Product Category")
plt.title(
    "Count of Delay and No Delay Orders",
    fontsize=14,
    fontweight="bold",
    color="green",
)
````

````python
# Plot the count of delay and no delay orders
plt.style.use('fivethirtyeight')
plt.subplots(figsize=(8, 5))
sns.countplot(data=sales, x="delay status",
             hue="Ship Mode")
plt.title(
    "Count of Delay and No Delay Orders",
    fontsize=14,
    fontweight="bold",
    color="green",
)
````

## Box plot
box plot is used to identify the distribution of numeric variables.

````python
sns.boxplot?
````

````python
# Drop ID/index columns so they don't distort the box plot scale
sales_subset = sales.drop(
    columns=["Order ID", "Unnamed: 0"], errors="ignore"
)

plt.subplots(figsize=(14, 5))
sns.boxplot(data=sales_subset)  # Plots remaining numeric columns
plt.xticks(rotation=45)
plt.show()
````

````python
plt.subplots(figsize=(14,5))
sns.boxplot(data=sales,
           x="Shipping Cost",
           y="Ship Mode") # gives a categorical coloumn for the y parameters
````

## HeatMap

heatmap is generally used in weather forcasting to find the hotspots in the map.here, we can used to recognize extreme values using colour 


````python
plt.figure(figsize=(8, 5))

sns.heatmap(
    pd.crosstab(sales["Region"], sales["delay status"]),
    cmap="spring",
    annot=True,
    fmt=".0f",
)

plt.title(
    "Delay Status by Region", fontsize=12, fontweight="bold", pad=10
)
plt.show()
````

# line plot

````python
dowjones = sns.load_dataset("dowjones")
dowjones.head()          # stock market data
````

````python
sns.lineplot(data=dowjones,x="Date", y="Price", color="green")
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=dowjones,
    x="Date",
    y="Price",
    color="green",
    linewidth=3,
    linestyle="--",
    marker="o",
    markersize=6
)

plt.title("Dow Jones Stock Price", fontsize=18, fontweight="bold")
plt.xlabel("Date", fontsize=14)
plt.ylabel("Price", fontsize=14)

plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

plt.figure(figsize=(14,6))

sns.lineplot(
    data=dowjones,
    x="Date",
    y="Price",
    color="#2E8B57",
    linewidth=3,
    marker="o",
    markersize=5
)

plt.title("Dow Jones Price Over Time", fontsize=18, weight="bold")
plt.xlabel("Date", fontsize=13)
plt.ylabel("Price ($)", fontsize=13)

plt.xticks(rotation=45)
plt.grid(alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,6))

sns.lineplot(data=dowjones, x="Date", y="Price", color="green")

plt.fill_between(
    dowjones["Date"],
    dowjones["Price"],
    color="green",
    alpha=0.3
)

plt.show()
````

````python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=dowjones,
    x="Date",
    y="Price",
    color="green",
    linewidth=2,
   
)

plt.fill_between(
    dowjones["Date"],
    dowjones["Price"],
    color="green",
    alpha=0.2
)

plt.title("Dow Jones Price")
plt.grid(True, alpha=0.3)

plt.show()
````

# Scatter Plot (`plt.scatter()`)

## What is a Scatter Plot?
A scatter plot displays individual data points to show the relationship between **two numerical variables**.

---

## When Should You Use It?

Use a scatter plot when you want to:
- Find the relationship between two variables.
- Identify trends or patterns.
- Detect outliers.
- Check for correlation (positive, negative, or no correlation).
- Visualize the distribution of data points.

---

## Syntax

```python
plt.scatter(x, y)
```

---

## Common Parameters

| Parameter | Description |
|-----------|-------------|
| `x` | Values for the x-axis |
| `y` | Values for the y-axis |
| `s` | Size of the markers |
| `c` | Color of the markers |
| `marker` | Shape of the markers |
| `alpha` | Transparency (0 to 1) |
| `edgecolor` | Border color of markers |
| `label` | Label for legend |

---

## Best For

- Correlation analysis
- Outlier detection
- Trend visualization
- Comparing two numerical features

---

## Example Use Cases

- Study Hours vs Exam Score
- Age vs Salary
- Height vs Weight
- Advertising Cost vs Sales
- Temperature vs Ice Cream Sales

---

## Advantages

- Easy to identify relationships.
- Helps detect outliers.
- Simple to understand.
- Useful before building Machine Learning models.

---

## Limitations

- Not suitable for categorical data.
- Can become cluttered with very large datasets.
- Doesn't show exact frequencies of overlapping points.

---

## Summary

- **Shows:** Relationship between two numerical variables.
- **Best for:** Correlation, trends, and outlier detection.
- **Common Function:** `plt.scatter()`

````python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")

iris.head()
````

````python
sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width"
)

plt.show()
````

````python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species",
    style="species",
    s=100,
    alpha=0.8
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.grid(alpha=0.3)

plt.show()
````

````python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species",
    style="species",
    s=100,
    alpha=0.8
)

plt.title("Petal Length vs Petal Width", fontsize=16, fontweight="bold")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.grid(alpha=0.3)
plt.legend(title="Species")

sns.despine()
plt.tight_layout()
plt.show()
````
