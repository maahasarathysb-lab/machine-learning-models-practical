```python
5)Write code to test whether the given number lies between 1 and 10
```

```python
num = int(input("Enter a number: "))

if 1 <= num <= 10:
    print(f"{num} lies between 1 and 10")
else:
    print(f"{num} does not lie between 1 and 10")
```
