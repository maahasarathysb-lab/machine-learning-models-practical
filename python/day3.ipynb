## slicing 
### syntax
obJect/ variable [start index_end index_step index]

**string[start : stop : step]**

```python
a = "maahasarathy"
a[2:5]
a[2:5:2]
a[5:2:1]
```

if step index is positive , **start index must be lesser than the end index** to get a non empty string

```python
a[-5:-2:-1]
```

if step index is negative, **start index must be grater than the end index** to get a non empty string object

```python
a[5::-1]
```

```python
a[5:-19:-1]
```

if step index is negative , if end index isn't specified, **it'll default to lowest index -1**(in negative indexing)

```python
a[:10:-1]
a[-1:10:-1]
```

if step index is **negative**, if start index isn't specified, **it'll default to highest index** (any indexing)

```python
a[::-1] # it takes the highest value in the string; lowest value of the string
```

```python
a[-1::-1]
```

```python
a[-1:-19:-1]
```

```python
a[::1
]
```

```python
a[0::1]
```

```python
a[0:18:1]
```

if step indexing is **positive** , if the start index isn't specified , **it'll default to 0**.
if step index is **positive** if end index isn't specified, **it'll default to end index +1**

```python
a[8:2]
```
