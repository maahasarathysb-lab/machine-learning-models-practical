```python
# ── Keep asking operator until valid ──────────────────────────────
while True:
    calc = input("Enter operation (+, -, *, /): ")
    if calc in ['+', '-', '*', '/']:
        break                           #valid operator → exit loop
    print("Invalid operator! Try again.")

# ── Keep asking first number until valid integer ──────────────────
while True:
    try:
        a = int(input("Enter first number  : "))
        break                           #valid integer → exit loop
    except ValueError:
        print("Invalid! Enter integers only.")

# ── Keep asking second number until valid + no division by zero ───
while True:
    try:
        b = int(input("Enter second number : "))
        assert not (calc == '/' and b == 0), "Division by zero! Enter non-zero value."
        break                           #valid value → exit loop
    except ValueError:
        print("Invalid! Enter integers only.")
    except AssertionError as e:
        print("Error:", e)              # asks again if b == 0 with /

# ── Calculate using if/elif (no imports) ─────────────────────────
if calc == '+':
    result = a + b
elif calc == '-':
    result = a - b
elif calc == '*':
    result = a * b
elif calc == '/':
    result = a / b

print("Result :", result)
```
