```python
def gym_register(plan, *members, **charges):
    try:
        total = 0
        
        # Plan charges
        if plan.lower() == "monthly":
            plan_fee = 1000
        elif plan.lower() == "yearly":
            plan_fee = 10000
        else:
            raise ValueError("Invalid plan. Choose 'monthly' or 'yearly'")
        
        # Calculate extra charges
        for fee in charges.values():
            total += fee
        
        # Total members
        member_count = len(members)
        
        total_charges = (plan_fee + total) * member_count
        
        print("Plan:", plan)
        print("Members:", members)
        print("Extra Charges:", charges)
        print("Total Members:", member_count)
        
        return total_charges
    
    except ValueError as v:
        print("Value Error:", v)
    except TypeError:
        print("Type Error: Charges must be numeric")
    except Exception as e:
        print("Unexpected Error:", e)
```

```python
def fruit_order(customer, *fruits, **prices):
    try:
        print("Customer Name:", customer)
        print("Fruits Ordered:")

        total = 0

        for fruit in fruits:
            fruit = fruit.strip().lower()
            print("-", fruit)

            try:
                if fruit in prices:
                    total += prices[fruit]
                else:
                    print(f"  Price for {fruit} not provided")
            except Exception as e:
                print("Error while processing fruit:", fruit)

        print("Total Amount:", total)

    except Exception as e:
        print("Something went wrong:", e)


```
